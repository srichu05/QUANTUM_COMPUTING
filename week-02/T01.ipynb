# Pauli-X Gate and Statevector Representation

## 1. Problem Statement
Create a 1-qubit quantum circuit initialized in the computational basis state $|0\rangle$, apply a Pauli-X ($X$) gate to the qubit, and obtain and display the resulting statevector.

---

## 2. Theoretical Background & Concepts

### Default Initial State
In quantum computing and Qiskit, qubits are initialized by default in the computational basis ground state $|0\rangle$, represented in column vector notation as:

$$|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$$

### The Pauli-X Gate (Quantum Bit-Flip)
The **Pauli-X gate** is a fundamental single-qubit unitary gate that acts as the quantum analogue of the classical NOT gate. Represented by the Pauli-X matrix:

$$X = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}$$

Applying the Pauli-X gate to the computational basis states causes a bit-flip transformation:

$$X|0\rangle = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} \begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} 0 \\ 1 \end{pmatrix} = |1\rangle$$

$$X|1\rangle = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} \begin{pmatrix} 0 \\ 1 \end{pmatrix} = \begin{pmatrix} 1 \\ 0 \end{pmatrix} = |0\rangle$$

### Statevector Representation
A **statevector** $|\psi\rangle$ is a complex-valued unit vector in a $2^n$-dimensional Hilbert space that provides a complete mathematical description of the pure state of an $n$-qubit quantum system. For a single qubit ($n = 1$):

$$|\psi\rangle = \alpha|0\rangle + \beta|1\rangle = \begin{pmatrix} \alpha \\ \beta \end{pmatrix}, \quad \text{where } |\alpha|^2 + |\beta|^2 = 1$$

The complex coefficients $\alpha$ and $\beta$ represent the probability amplitudes for measuring states $|0\rangle$ and $|1\rangle$, respectively.

For this circuit, after applying the Pauli-X gate to $|0\rangle$, the resulting statevector is:

$$|\psi\rangle = |1\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix} = 0|0\rangle + 1|1\rangle$$

which corresponds to an amplitude of $0$ for $|0\rangle$ and an amplitude of $1$ for $|1\rangle$.

---

## 3. Circuit Flow
The sequence of operations in this quantum circuit is:

$$|0\rangle \xrightarrow{\quad \text{Pauli-X Gate} \quad} |1\rangle$$

The qubit begins in the computational ground state $|0\rangle$, is transformed by the Pauli-X gate into the excited state $|1\rangle$, and is then directly inspected via its statevector.

## 4. Tools & Libraries Used
- **Qiskit**: IBM's open-source quantum computing SDK for composing and displaying quantum circuits (`QuantumCircuit`).
- **`qiskit.quantum_info.Statevector`**: Analytical module in Qiskit for computing and inspecting exact quantum statevectors from circuit instructions without introducing measurements or shot sampling.

### Step 1: Import Required Libraries
Import `QuantumCircuit` to build the quantum circuit and `Statevector` from `qiskit.quantum_info` to inspect the exact quantum statevector.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

### Step 2: Create the Quantum Circuit
Initialize a 1-qubit quantum circuit. In the computational basis, the qubit is initialized in state $|0\rangle$ by default.

```python
# Create a 1-qubit circuit
qc = QuantumCircuit(1)
```

### Step 3: Apply the Pauli-X Gate
Apply the Pauli-X gate (`qc.x(0)`) to qubit 0 to execute the quantum bit-flip operation.

```python
# Apply X gate
qc.x(0)
```

### Step 4: Obtain the Statevector
Extract the exact quantum statevector directly from the circuit instructions using `Statevector.from_instruction(qc)`.

```python
# Obtain statevector
state = Statevector.from_instruction(qc)
```

### Step 5: Display the Circuit and Statevector
Print the text-based circuit diagram and inspect the resulting statevector amplitudes.

```python
print("Quantum Circuit:")
print(qc)

print("\nStatevector:")
print(state)
```

## Conclusion

- **State Transformation**: Applying the Pauli-X gate to a qubit initialized in $|0\rangle$ successfully performs a quantum bit-flip, transitioning the qubit into the orthogonal basis state $|1\rangle$.
- **Statevector Verification**: The computed statevector `Statevector([0.+0.j, 1.+0.j], dims=(2,))` confirms the theoretical transformation:
  $$|\psi\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix} = 0|0\rangle + 1|1\rangle$$
  The amplitude for $|0\rangle$ is $0$, and the amplitude for $|1\rangle$ is $1$, confirming a $100\%$ theoretical probability ($P(1) = |1|^2 = 1.0$) of finding the qubit in state $|1\rangle$.
- **Analytical State Inspection**: Using `Statevector.from_instruction(qc)` enables direct mathematical verification of unitary quantum gate transformations without the need for probabilistic measurement gates or finite shot sampling.
