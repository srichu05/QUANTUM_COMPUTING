## 2. Theoretical Background & Concepts

### The Pauli Gate Family
The Pauli operators ($X, Y, Z$) are fundamental single-qubit unitary and Hermitian gates. Together with the $2 \times 2$ identity matrix $I$, they form a basis for all single-qubit Hermitian operators and geometrically represent rotations by $\pi$ radians ($180^\circ$) around the $X$, $Y$, and $Z$ axes of the Bloch sphere.

### The Pauli-X Gate (Bit-Flip)
The **Pauli-X gate** acts as the quantum equivalent of a classical NOT gate by exchanging the computational basis states:

$$X = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}$$

$$X|0\rangle = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}\begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} 0 \\ 1 \end{pmatrix} = |1\rangle$$

$$X|1\rangle = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}\begin{pmatrix} 0 \\ 1 \end{pmatrix} = \begin{pmatrix} 1 \\ 0 \end{pmatrix} = |0\rangle$$

### The Pauli-Y Gate (Bit-Flip + Phase Shift)
The **Pauli-Y gate** combines a bit-flip operation with an imaginary phase factor of $\pm i$:

$$Y = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}$$

$$Y|0\rangle = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}\begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} 0 \\ i \end{pmatrix} = i|1\rangle$$

$$Y|1\rangle = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}\begin{pmatrix} 0 \\ 1 \end{pmatrix} = \begin{pmatrix} -i \\ 0 \end{pmatrix} = -i|0\rangle$$

The imaginary component ($0.+1.j$) in $Y|0\rangle$ directly reflects the complex phase factor $i = e^{i\pi/2}$ introduced by the Pauli-Y operation. While standard computational basis measurement probabilities remain $|i|^2 = 1.0$, the statevector explicitly captures this complex phase.

### The Pauli-Z Gate (Phase-Flip)
The **Pauli-Z gate** performs a phase-flip operation by introducing a relative phase of $\pi$ (a factor of $-1$) to the $|1\rangle$ component while leaving the $|0\rangle$ component unchanged:

$$Z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$$

$$Z|0\rangle = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}\begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} 1 \\ 0 \end{pmatrix} = |0\rangle$$

$$Z|1\rangle = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}\begin{pmatrix} 0 \\ 1 \end{pmatrix} = \begin{pmatrix} 0 \\ -1 \end{pmatrix} = -|1\rangle$$

Because the circuit starts in $|0\rangle$, applying $Z$ leaves the state unaltered ($Z|0\rangle = |0\rangle$).

### Statevector Representation
A **statevector** $|\psi\rangle$ is a complex unit vector containing the exact probability amplitudes of all computational basis states:

$$|\psi\rangle = \alpha|0\rangle + \beta|1\rangle = \begin{pmatrix} \alpha \\ \beta \end{pmatrix}, \quad |\alpha|^2 + |\beta|^2 = 1$$

Statevector analysis allows direct inspection of both the magnitudes (populations) and the complex phases of quantum states without introducing measurement collapse.

### Step 1: Import Required Libraries
Import `QuantumCircuit` to construct quantum circuits and `Statevector` from `qiskit.quantum_info` to extract exact quantum statevectors.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

### Step 2: Define the Single-Qubit Gate Experiment
Define a reusable function `display_gate(gate)` that initializes a 1-qubit circuit in $|0\rangle$, applies the specified Pauli gate ($X$, $Y$, or $Z$), renders the text circuit diagram, and prints the resulting statevector.

```python
def display_gate(gate):
    qc = QuantumCircuit(1)

    if gate == "X":
        qc.x(0)
    elif gate == "Y":
        qc.y(0)
    elif gate == "Z":
        qc.z(0)

    print(f"\n----- {gate} Gate -----")
    print(qc.draw())
    print(Statevector.from_instruction(qc))
```

### Step 3: Execute Gate Transformations
Iterate through the list `["X", "Y", "Z"]` and call `display_gate()` for each gate to observe the circuit diagram and resulting statevector.

```python
for gate in ["X", "Y", "Z"]:
    display_gate(gate)
```

## Statevector Interpretation and Analysis

### Analyze the Pauli-X Gate
- **Input State**: $|0\rangle = [1.+0.j, 0.+0.j]^T$
- **Output Statevector**: `Statevector([0.+0.j, 1.+0.j], dims=(2,))`
- **Transformation**: $X|0\rangle = |1\rangle$. The Pauli-X gate acts as a complete bit-flip, transferring $100\%$ of the probability amplitude from $|0\rangle$ to $|1\rangle$.

---

### Analyze the Pauli-Y Gate
- **Input State**: $|0\rangle = [1.+0.j, 0.+0.j]^T$
- **Output Statevector**: `Statevector([0.+0.j, 0.+1.j], dims=(2,))`
- **Transformation**: $Y|0\rangle = i|1\rangle$. The Pauli-Y gate performs a bit-flip while introducing an imaginary phase factor of $+i = e^{i\pi/2}$. The amplitude of $|1\rangle$ is purely imaginary ($0.+1.j$), which is uniquely visible through statevector inspection.

---

### Analyze the Pauli-Z Gate
- **Input State**: $|0\rangle = [1.+0.j, 0.+0.j]^T$
- **Output Statevector**: `Statevector([1.+0.j, 0.+0.j], dims=(2,))`
- **Transformation**: $Z|0\rangle = |0\rangle$. Because the Pauli-Z gate acts as a phase-flip specifically on the $|1\rangle$ state ($Z|1\rangle = -|1\rangle$), it leaves the ground state $|0\rangle$ completely unaffected.

## Conclusion

- **Distinct Pauli Transformations**: The experiment clearly demonstrates the distinct operational characteristics of the three Pauli gates when acting on the initial state $|0\rangle$:
  - The **Pauli-X gate** executes a bit-flip ($|0\rangle \to |1\rangle$).
  - The **Pauli-Y gate** executes a combined bit-flip and complex phase rotation ($|0\rangle \to i|1\rangle$).
  - The **Pauli-Z gate** acts as a phase-flip that leaves $|0\rangle$ invariant ($|0\rangle \to |0\rangle$).
- **Value of Statevector Analysis**: Direct statevector inspection makes both amplitude magnitudes and complex phase factors explicitly visible, revealing the imaginary component ($0.+1.j$) of the Pauli-Y gate that would otherwise be obscured in standard computational basis population measurements.
