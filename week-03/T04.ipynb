# Quantum Coin Flip Using a Hadamard Gate

## 1. Objective
The objective of this experiment is to demonstrate how a single qubit can model an idealized quantum coin flip by applying a Hadamard gate to an initial ground state $|0\rangle$ and calculating the resulting computational-basis probabilities.

---

## 2. Concept & Theoretical Background

### Initial State and the Hadamard Gate
In quantum computing, a single qubit is initialized by default in the computational ground state:

$$|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$$

Applying a **Hadamard gate** ($H$) transforms $|0\rangle$ into an equal superposition of the two computational-basis states:

$$H = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}$$

$$H|0\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}} = |+\rangle$$

In this state, both basis states $|0\rangle$ and $|1\rangle$ share an equal probability amplitude of $\frac{1}{\sqrt{2}} \approx 0.7071$.

### Mapping to a Quantum Coin Flip
The two orthogonal computational basis states can be directly mapped to the two binary outcomes of a classical coin flip:
- $|0\rangle \longrightarrow \text{Heads}$
- $|1\rangle \longrightarrow \text{Tails}$

According to the **Born rule**, the probability of observing each outcome upon measurement is given by the squared magnitude of its amplitude:

$$P(\text{Heads}) = P(0) = \left|\frac{1}{\sqrt{2}}\right|^2 = \frac{1}{2} = 50\%$$

$$P(\text{Tails}) = P(1) = \left|\frac{1}{\sqrt{2}}\right|^2 = \frac{1}{2} = 50\%$$

For an ideal Hadamard operation, both outcomes have an exact theoretical probability of 50%.

---

## 3. Circuit Flow
The transformation sequence is:

$$|0\rangle \xrightarrow{\quad \text{Hadamard Gate (H)} \quad} |+\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}} \xrightarrow{\quad \text{Statevector Probabilities} \quad} P(\text{Heads}) = 50\%, \quad P(\text{Tails}) = 50\%$$

## 4. Implementation
- **Circuit Construction**: `coin = QuantumCircuit(1)` creates a one-qubit circuit initialized in the computational ground state $|0angle$.
- **Applying Hadamard**: `coin.h(0)` applies the Hadamard gate to the qubit to prepare the equal superposition state $|+angle$.
- **Statevector Calculation**: `Statevector.from_instruction(coin)` obtains the resulting quantum state directly from the circuit instructions without performing physical measurement collapse.
- **Probability Extraction**: `heads, tails = state.probabilities()` extracts the probabilities of the computational-basis states.
- **Percentage Conversion**: The probabilities are converted into percentages and displayed (`heads*100:.1f}%` and `tails*100:.1f}%`).

````python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

coin = QuantumCircuit(1)

coin.h(0)

state = Statevector.from_instruction(coin)

heads, tails = state.probabilities()

print("Quantum Coin Circuit")
print(coin.draw())

print(f"\nHeads : {heads*100:.1f}%")
print(f"Tails : {tails*100:.1f}%")
````
