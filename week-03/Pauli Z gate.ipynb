# Hadamard and Pauli-Z Phase Transformation

## 1. Objective
The objective of this experiment is to study the effect of applying a Hadamard gate ($H$) followed by a Pauli-Z gate ($Z$) to a single qubit initially prepared in the computational ground state $|0\rangle$, and to observe how the quantum state and its relative phase evolve without altering computational-basis measurement probabilities.

---

## 2. Concept & Theoretical Background

### Initial State and Hadamard Superposition
A single qubit starts in the default computational ground state:

$$|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$$

Applying a **Hadamard gate** ($H$) creates an equal linear superposition of the basis states:

$$H = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}$$

$$H|0\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}} = |+\rangle$$

In the $|+\rangle$ state, both computational basis states have equal positive amplitudes of $+1/\sqrt{2}$, with a relative phase of $\phi = 0$.

### The Pauli-Z Phase-Flip Gate
The **Pauli-Z gate** is a unitary phase-flip operator represented by:

$$Z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$$

When acting on the computational basis states:
- It leaves $|0\rangle$ unchanged: $Z|0\rangle = |0\rangle$
- It introduces a phase flip (a factor of $-1 = e^{i\pi}$) to $|1\rangle$: $Z|1\rangle = -|1\rangle$

### Combined Hadamard and Pauli-Z Transformation
Applying the Pauli-Z gate to the superposition state $|+\rangle$ yields:

$$|\psi\rangle = Z(H|0\rangle) = Z\left(\frac{|0\rangle + |1\rangle}{\sqrt{2}}\right) = \frac{|0\rangle - |1\rangle}{\sqrt{2}} = |-\rangle$$

In column vector notation:

$$|\psi\rangle = \begin{pmatrix} \frac{1}{\sqrt{2}} \\ -\frac{1}{\sqrt{2}} \end{pmatrix} \approx \begin{pmatrix} 0.70710678 \\ -0.70710678 \end{pmatrix}$$

### Relative Phase vs. Measurement Probabilities
- **Relative Phase**: The phase difference between $|0\rangle$ and $|1\rangle$ shifts from $0$ to $\pi$ radians ($180^\circ$), changing the state from $|+\rangle$ (along the $+X$ axis of the Bloch sphere) to $|-\rangle$ (along the $-X$ axis).
- **Measurement Probabilities**: According to the Born rule, the probability of measuring each computational basis state is given by the squared magnitude of its amplitude:
  $$P(0) = \left|\frac{1}{\sqrt{2}}\right|^2 = 0.5 \quad (50\%)$$
  $$P(1) = \left|-\frac{1}{\sqrt{2}}\right|^2 = 0.5 \quad (50\%)$$
Therefore, the relative phase changes while the computational-basis measurement probabilities remain strictly $0.5$ for both states.

## 3. Implementation
The circuit is implemented using Qiskit's `QuantumCircuit` on a single qubit. The gate sequence $[H, Z]$ is defined as a list of gate methods `operations = [qc.h, qc.z]` and applied iteratively using a loop. 

Rather than introducing measurement collapse or sampling noise via an AerSimulator, `Statevector.from_instruction(qc)` is used to calculate the exact statevector and theoretical probabilities directly from the circuit instructions.

````python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)

operations = [qc.h, qc.z]

for op in operations:
    op(0)

state = Statevector.from_instruction(qc)

print("Circuit")
print(qc.draw())

print("\nFinal State")
print(state)

print("\nProbabilities")
print(state.probabilities())
````

## 4. Results & Observations

### Observed Output Analysis
- **Quantum Circuit Diagram**:
  Displays a 1-qubit quantum register subjected sequentially to a Hadamard gate ($H$) followed by a Pauli-Z gate ($Z$):
  ```
  q: ┤ H ├┤ Z ├
  ```

- **Final Statevector**:
  $$\text{Statevector}([0.70710678+0.\text{j}, -0.70710678+0.\text{j}], \text{dims}=(2,))$$
  where $0.70710678 \approx \frac{1}{\sqrt{2}}$. The positive real amplitude $+0.70710678$ corresponds to state $|0\rangle$, while the negative real amplitude $-0.70710678$ corresponds to state $|1\rangle$.

- **Theoretical Probabilities**:
  $$\text{Probabilities: } [0.5, 0.5]$$
  Both basis states $|0\rangle$ and $|1\rangle$ have identical probabilities of $0.5$ ($50\%$).

- **Phase Significance**:
  The negative sign in the $|1\rangle$ amplitude represents the relative phase factor $-1 = e^{i\pi}$ introduced by the Pauli-Z gate. This phase inversion rotates the statevector from $|+\rangle$ to $|-\rangle$ without altering the computational basis probabilities.

## 5. Conclusion
This experiment clearly demonstrates that quantum gates can modify the relative phase of a quantum superposition without changing its computational-basis measurement probabilities. While standard projective measurements in the $Z$-basis only reveal outcome probabilities ($50/50$), statevector analysis preserves the full phase information, confirming that the quantum state has been transformed from $|+angle$ into $|-\rangle$.
