```python
import matplotlib.pyplot as plt

n_values = list(range(2, 11))

classical_queries = []
quantum_queries = []
speedup = []

print("\nSIMON'S ALGORITHM QUERY COMPLEXITY")
print("=" * 70)
print("n    Classical Queries    Quantum Queries    Speedup")
print("-" * 70)

for n in n_values:
    classical = 2 ** n
    quantum = n - 1
    ratio = classical / quantum

    classical_queries.append(classical)
    quantum_queries.append(quantum)
    speedup.append(ratio)

    print(
        f"{n:<4}"
        f"{classical:<21}"
        f"{quantum:<19}"
        f"{ratio:.2f}x"
    )

print("-" * 70)

plt.figure(figsize=(10, 6))

plt.plot(
    n_values,
    classical_queries,
    marker="o",
    linewidth=2,
    label="Classical brute force"
)

plt.plot(
    n_values,
    quantum_queries,
    marker="s",
    linewidth=2,
    label="Simon's algorithm"
)

plt.xlabel("Secret String Length n")
plt.ylabel("Number of Oracle Queries")
plt.title("Classical vs Quantum Query Complexity")

plt.yscale("log")

plt.xticks(n_values)

plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))

plt.plot(
    n_values,
    speedup,
    marker="o",
    linewidth=2
)

plt.xlabel("Secret String Length n")
plt.ylabel("Classical / Quantum Query Ratio")
plt.title("Simon Algorithm Query Speedup")

plt.xticks(n_values)

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
